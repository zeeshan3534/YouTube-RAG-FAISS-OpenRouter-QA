```python
# All imports
from urllib.parse import urlparse, parse_qs
from youtube_transcript_api import YouTubeTranscriptApi
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings
from openai import OpenAI
import faiss
import numpy as np
import requests
import json
```

```python
# youtube Id Extractor

def get_youtube_video_id(url: str) -> str | None:
    parsed = urlparse(url)

    if parsed.hostname in ("youtu.be", "www.youtu.be"):
        return parsed.path.strip("/").split("/")[0] or None

    if parsed.hostname in ("youtube.com", "www.youtube.com", "m.youtube.com"):
        if parsed.path == "/watch":
            return parse_qs(parsed.query).get("v", [None])[0]

        if parsed.path.startswith("/shorts/"):
            return parsed.path.split("/")[2]

        if parsed.path.startswith("/embed/"):
            return parsed.path.split("/")[2]

    return None
```

```python
# YouTube URL
url = "https://youtu.be/Qtl8lJwbd4g?si=F_tMOiAXZw-mmHO6"

# Extract video ID
video_id = get_youtube_video_id(url)

print("Video ID:", video_id)

try:
    # Create API instance
    api = YouTubeTranscriptApi()

    # Fetch transcript
    transcript = api.fetch(video_id)

    # Print transcript
    join_transcript = " ".join(entry.text for entry in transcript)
    # for i in transcript:
    #   print(i.text)
    print(join_transcript)


except Exception as e:
    print(f"An error occurred: {e}")
```

```python
from requests.models import CONTENT_CHUNK_SIZE
# Text Splitting
splitter = RecursiveCharacterTextSplitter(chunk_size = 1000,chunk_overlap = 200)
chunk = splitter.create_documents([join_transcript])
print(chunk[0])
```

```python
len(chunk)
```

```python
chunks = [doc.page_content for doc in chunk]
print(chunks)
```

```python
#api key
api_key = "[YOUR-API-KEY]"
```

```python
# Response initialization

response = requests.post(
  url="https://openrouter.ai/api/v1/embeddings",
  headers={
    "Authorization": f"Bearer {api_key}",
    "Content-Type": "application/json",

  },
  data=json.dumps({
    "model": "liquid/lfm-2.5-embedding-350m:free",
    "input": chunks,
    "encoding_format": "float"
  })
)
```

```python
print(response.content)
```

```python
result = response.json()
embeddings = [item["embedding"] for item in result["data"]]
```

```python

print("Number of embeddings:", len(embeddings))
print("Embedding dimensions:", len(embeddings[0]))
```

```python
# initializing the vector store and storing data
vectors = np.array(embeddings,dtype=np.float32)
dimensions = vectors.shape[1]
index = faiss.IndexFlatL2(dimensions)
index.reset()
print(index.ntotal)
index.add(vectors)
```

```python
print("Vectors stored:", index.ntotal)
```

```python
#User Query
query = "is this video talking about mercedes-benz if yes can you give me the summary of this video"

new_api_key = "[Your-LLM-API-KEY-HERE]"
# create embedding of the question

query_response = requests.post(
  url="https://openrouter.ai/api/v1/embeddings",
  headers={
    "Authorization": f"Bearer {new_api_key}",
    "Content-Type": "application/json",

  },
  data=json.dumps({
    "model": "liquid/lfm-2.5-embedding-350m:free",
    "input": query,
    "encoding_format": "float"
  })
)

embedding_query = query_response.json()["data"][0]["embedding"]
print(embedding_query)
```

```python
query_vector = np.array(
    [embedding_query],
    dtype=np.float32
)
```

```python
k = 3

distances, indices = index.search(query_vector, k)
```

```python
print(indices)
relevant_chunks = [chunks[i] for i in indices[0]]
join_chuncks = " ".join(item for item in relevant_chunks)
print(join_chuncks)
```
